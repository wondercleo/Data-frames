```python
# CREATING DATA FRAMES IN PANDAS

# creating from a dictionary

data = {
    "name": ["John","Mary","peter"],
    "age": [25, 30, 22],
    "city": ["Embu", "Nairobi", "Mombasa"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Creating from list of lists

data = [
    ["John", 25, "Embu"],
    ["Mary", 30, "Nairobi"],
    ["Peter", 22, "Mombasa"]
]
df = pd.DataFrame(data, columns=["Name","Age","County"])
print(df)
```

```python
# creating an empty data frame
df = pd.DataFrame()
print(df)
```

```python
# creating the columns first in a dataframe
df = pd.DataFrame(columns=["Name","Age","County"])
print(df)
```

```python
# creating from a dataframe from numpy
data = np.array([
    [10, 20, 30],
    [40, 50, 60]])
data = pd.DataFrame(data)
print(data)
```

```python
# adding a column in a numpy array
data = np.array([
    [10, 20, 30],
    [40, 50, 60]])
data = pd.DataFrame(data, columns=["number","age","amount"])
print(data)
```

```python
# adding indexes in numpy dataframes
data = np.array([
    [10, 20, 30],
    [40, 50, 60]])
data = pd.DataFrame(data, columns=["number","age","amount"],index=["first","last"])
print(data)
```

```python
# creating a dataframe with an index and column already
data = [
    [80, 90, 70],
    [60, 75, 85],
    [25, 30, 22]
]

df = pd.DataFrame(
    data,
    index=["math", "english", "age"],
    columns=["John", "Mary", "Peter"]
)

print(df)
```
